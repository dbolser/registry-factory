# Register Objects in two ways

```python
from registry_factory.registry import Registry
```

```python
@Registry.register("add")
def add(a: int, b: int) -> int:
    return a + b

class String:
    def __init__(self, value: str) -> None:
        self.value = value

    def __str__(self):
        return self.value

Registry.register_prebuilt(String, "string")
```

```python
print(Registry.keys())
print(Registry.values())
print(Registry.items()) # items also shows additional parameters for calling (i.e., version etc.)
```

## Calling registered objects

```python
string = Registry.get("string")
string = string("hey")

print(string)
```
